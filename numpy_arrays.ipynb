# WAYS TO MAKE ARRAYS

```python
int_array = np.array([1,4,2,5,6])

print(int_array)
```

creating an array is straightforward

```python
mixed_array = np.array([3.14,20,52.2,10])

print(mixed_array)
```

If a numpy array contains both integers and doubles, all integers are made into doubles by default. You can explicitly set the data type of the resulting array however, which I'll cover here.

```python
explicit_set_array = np.array([1,4,2,5,6], dtype=np.float64)

print(explicit_set_array)
```

If you want to explicitly set the type of an array, you can use the `dtype` keyword.

```python
arr_multidim = np.array([range(i, i + 3) for i in [2, 4, 6]])

print(arr_multidim)
```

Unlike python lists, NumPy arrays can be multidimensional. Above is a two-dimensional array, where each inner list is its own row.

```python
zero_arr = np.zeros((3,2),dtype=int)

print(zero_arr)
```

You can create an array filled with zeroes by using np.zeros(quantity).

```python
ones_arr = np.ones((3,5),dtype=int)

print(ones_arr)
```

You can create an array filled with ones by using np.ones(quantity). Behaves exactly like np.zeros otherwise.

```python
osix_arr = np.full((7,12),2006)

print(osix_arr)
```

You can create an array filled with a specific number by using np.full(quantity, num). Behaves exactly like np.zeros otherwise.

```python
arange_arr = np.arange(0,20,2)

print(arange_arr)
```

You can create an array filled with a linear sequence using np.arange. This one starts at 0 (arg 1), ends at 20 (arg 2), in steps of 2 (arg 3).

```python
np.linspace(0,1,5)
```

You can create an array of #arg3 evenly spaced values between arg 1 and arg 2 using np.linspace.

```python
rand_arr = np.random.random((3,3))

print(rand_arr)
```

You can create an array of random values using np.random.random().

```python
arr_randNorm = np.random.normal(0,1,(3,3))

print(arr_randNorm)
```

```python
arr_randInt = np.random.randint(0,10,(3,3))

print(arr_randInt)
```

# ARRAYS AND THEIR PROPERTIES

```python
np.random.seed(0) # seed for reproducibility
```

```python
arr_1d = np.random.randint(10,size=6)

arr_2d = np.random.randint(10,size=(3,4))

arr_3d = np.random.randint(10, size=(3,4,5))
```

Each array has attributes, including:

`ndim` - The number of dimensions

`shape` - The size of each dimension

`size` - The total size of the array

`dtype` - The data type of the array

`itemsize` - The size of each array element in bytes

`nbytes` - The total size of the array in bytes

> Generally, `nbytes` should be equal to `itemsize * size`.

```python
print("arr_3d ndim: ", arr_3d.ndim)
print("arr_3d shape:", arr_3d.shape)
print("arr_3d size: ", arr_3d.size)
print("arr_3d dtype: ", arr_3d.dtype)
print("arr_3d itemsize: ", arr_3d.itemsize)
print("arr_3d nbytes: ", arr_3d.nbytes)
```

Indexing in numpy is very similar to standard Python.

you know how to do this already.

In a multi-dimensional array, items can be accessed using a comma separated indice tuple (eg. arr_3d[3,2])

NumPy arrays have a fixed type however, so any doubles will be converted to integers and vice versa.

Likewise, slicing is also very similar.

Slicing multidimensionally can be done very similarly, with multiple slices being done via commas.


# HOWEVER!

Array slices return views, rather than copies of the array data. To *copy*, just put .copy() at the end.

```python
x = np.array([1,2,3])
y = np.array([3,2,1])

np.concatenate([x,y])
```

np.concatenate allows you to join two or more arrays in NumPy, and it can also be used for 2D arrays.

```python
grid = np.concatenate([arr_2d,arr_2d])

print(grid)
```

You can also designate which axis the function wil concatenate along.

```python
grid = np.concatenate([arr_2d,arr_2d], axis=1)

print(grid)
```

When working with arrays of mixed dimensions, it can be better to use np.vstack and np.hstack functions.

Splitting can be done using np.split, np.hsplit, and np.vsplit.

# VECTORIZATION DEMO

```python
def loop_sum(arr):
    total = 0
    for idx in arr:
        total += idx

    return total

def vectorized_sum(arr):
    return np.sum(arr)
```

```python

arr = np.random.rand(100_000_000)

start = time.time()
loop_sum(arr)
loop_speed = time.time() - start
print(f"Loop: {loop_speed:.4f}s")


start = time.time()
vectorized_sum(arr)
vectorized_speed = time.time() - start
print(f"Vectorized: {vectorized_speed:.4f}s")

print(f"Speedup: {loop_speed/vectorized_speed:.1f}x")
```

Vectorized is about 80-180x times faster than looping, especially with larger values.
